```python
import pandas as pd

df = pd.read_csv("Amazon.csv")

print(df.isnull().sum())

df.drop_duplicates(inplace=True)

df['OrderDate'] = pd.to_datetime(df['OrderDate'])

df.dropna(subset=['OrderID', 'OrderDate', 'ProductName', 
                  'Category', 'Quantity', 'TotalAmount'], inplace=True)

df['Category'] = df['Category'].str.strip()
df['Brand'] = df['Brand'].str.strip()
df['ProductName'] = df['ProductName'].str.strip()

print(df.shape)
print(df.head())
```

```python
df.to_csv("Amazon_Cleaned.csv", index=False)
```

```python
category_sales = df.groupby('Category')['TotalAmount'].sum()
print(category_sales)
```

```python
total_sales = df['TotalAmount'].sum()
print("Total Sales:", total_sales)
```

```python
product_sales = df.groupby('ProductName')['TotalAmount'].sum()
print(product_sales.sort_values(ascending=False).head(10))
```

```python
brand_sales = df.groupby('Brand')['TotalAmount'].sum()
print(brand_sales.sort_values(ascending=False).head(10))
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
monthly_sales = df.groupby(df['OrderDate'].dt.to_period('M'))['TotalAmount'].sum()
print(monthly_sales)
```

```python
category_quantity = df.groupby('Category')['Quantity'].sum()

print(category_quantity.sort_values(ascending=False))
```

```python
average_order_value = df['TotalAmount'].mean()

print("Average Order Value:", average_order_value)
```
